```python
import numpy as np
```

```python
employee_id = np.arange(1001, 1031)

age = np.array([
    23, 28, 35, 25, 32, 29, 40, 26, 31, 24,
    38, 27, 33, 30, 22, 36, 28, 34, 25, 39,
    29, 31, 26, 37, 24, 32, 35, 27, 30, 23
])

department = np.array([
    "IT", "HR", "Sales", "IT", "Finance", "Sales",
    "IT", "HR", "Finance", "IT", "Sales", "IT",
    "HR", "Finance", "IT", "Sales", "HR", "IT",
    "Finance", "Sales", "IT", "Finance", "HR",
    "Sales", "IT", "HR", "Finance", "IT", "Sales", "HR"
])

salary = np.array([
    45000, 35000, 55000, 60000, 50000, 42000,
    80000, 38000, 65000, 48000, 72000, 52000,
    40000, 58000, 32000, 75000, 46000, 90000,
    43000, 68000, 54000, 62000, 37000, 85000,
    41000, 49000, 70000, 56000, 63000, 39000
])

experience = np.array([
    1, 3, 8, 4, 6, 2, 12, 3, 7, 2,
    10, 5, 4, 8, 1, 11, 3, 13, 2, 9,
    6, 7, 2, 12, 3, 5, 9, 6, 8, 1
])

performance = np.array([
    3.5, 4.2, 4.5, 3.8, 4.1, 3.2, 4.8, 3.9, 4.6, 3.4,
    4.7, 4.0, 3.6, 4.3, 2.8, 4.9, 3.7, 4.8, 3.1, 4.6,
    4.2, 3.9, 3.3, 4.7, 3.5, 4.1, 4.4, 3.8, 4.5, 3.0
])

working_hours = np.array([
    40, 38, 45, 42, 40, 36, 48, 39, 44, 37,
    46, 41, 38, 43, 35, 47, 40, 50, 36, 45,
    42, 41, 37, 49, 38, 40, 44, 42, 46, 36
])

sales = np.array([
    12000, 8000, 25000, 18000, 15000, 11000,
    32000, 9000, 22000, 13000, 28000, 17000,
    10000, 20000, 7000, 30000, 14000, 35000,
    9000, 27000, 19000, 21000, 8500, 33000,
    11000, 16000, 24000, 18000, 26000, 7500
])
```

```python
total_employees = np.size(employee_id)
print("Total number of employees:", total_employees)
```

```python
first_10_ids = employee_id[0:10]
print("First 10 Employee IDs:", first_10_ids)
```

```python
highest_salary = np.max(salary)
lowest_salary = np.min(salary)

print("Highest Salary:", highest_salary)
print("Lowest Salary:", lowest_salary)
```

```python
average_salary = np.mean(salary)
print("Average Salary:", average_salary)
```

```python
high_salary_employees = employee_id[salary > 50000]
print("Employees with salary > 50000:", high_salary_employees)
```

```python
age_filter = (age >= 25) & (age <= 35)
employees_age_25_35 = employee_id[age_filter]

print("Employees with age 25 to 35:", employees_age_25_35)
```

```python
condition = (experience > 5) & (performance >= 4.0)
result = employee_id[condition]

print("Employees with exp > 5 and performance >= 4.0:", result)
```

```python
total_sales = np.sum(sales)
print("Total Sales:", total_sales)
```

```python
salary_label = np.where(salary > 60000, "High Salary", "Normal Salary")

for i in range(len(employee_id)):
    print(employee_id[i], "-", salary[i], "-", salary_label[i])
```

```python
new_salary = np.where(salary < 40000, salary * 1.10, salary)
print("New Salary:", new_salary)
```

```python
perf_condition = (performance >= 3.5) & (performance <= 4.5)
result = employee_id[perf_condition]

print("Employees with performance 3.5 to 4.5:", result)
```

```python
median_salary = np.median(salary)
print("Median Salary:", median_salary)
```

```python
condition = (working_hours > 40) | (sales > 25000)
result = employee_id[condition]

print("Employees with hours > 40 OR sales > 25000:", result)
```

```python
exp_label = np.where(experience >= 5, "Experienced", "Fresher")

for i in range(len(employee_id)):
    print(employee_id[i], "-", exp_label[i])
```

```python
std_salary = np.std(salary)
print("Standard Deviation:", std_salary)
```

```python
avg_salary = np.mean(salary)
condition = salary > avg_salary
result = employee_id[condition]

print("Average Salary:", avg_salary)
print("Employees earning more than average:", result)
print("Salaries:", salary[condition])
```

```python
sqrt_sales = np.sqrt(sales)
for i in range(len(employee_id)):
    emp_id = employee_id[i]
    s = sales[i]
    sq = sqrt_sales[i]
    print(f"{emp_id} - Sales: {s} - Sqrt: {sq:.2f}")
```

```python
condition = department != "IT"
result = employee_id[condition]

print("Employees not in IT:", result)
```

```python
bonus = np.where(performance >= 4.5, 5000, 2000)
for i in range(len(employee_id)):
    print(f"{employee_id[i]} - Score: {performance[i]} - Bonus: {bonus[i]}")
```

```python
highest = np.max(salary)
lowest = np.min(salary)
difference = highest - lowest

print("Highest Salary:", highest)
print("Lowest Salary:", lowest)
print("Difference:", difference)
```
