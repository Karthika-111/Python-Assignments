```python
product_list = ["laptop", "mouse", "keyboard", "headphones", "monitor"]
print("Available Products:", product_list)

product_tuple = (
    ("laptop", 40000),
    ("mouse", 500),
    ("keyboard", 1500),
    ("headphones", 2000),
    ("monitor", 10000)
)

product_category = ["electronics", "accessories", "accessories", "audio", "electronics"]

cart = {}

categories_set = set()

while True:
    name = input("\nEnter product name: ")
    name = name.lower().strip()

    if name not in product_list:
        print("Error: Product unavailable")
        continue

    qty = int(input("Enter quantity: "))

    if qty <= 0:
        print("Error: Quantity should be more than 0")
        continue

    if name in cart:
        cart[name] = cart[name] + qty
    else:
        cart[name] = qty

    index = 0
    for i in range(len(product_list)):
        if product_list[i] == name:
            index = i
            break
    categories_set.add(product_category[index])

    print(f"Added {qty} {name} to cart")

    choice = input("Do you want to continue shopping? (yes/no): ")
    choice = choice.lower().strip()

    if choice == "no" or choice == "n" or choice == "2":
        break

if len(cart) == 0:
    print("\nCart is empty")
else:
    total = 0
    print("\n--- Final Bill ---")
    for product_name in cart:
        product_qty = cart[product_name]
        product_price = 0
        for item in product_tuple:
            if item[0] == product_name:
                product_price = item[1]
        
        bill = product_price * product_qty
        print(product_name, "x", product_qty, "=", bill)
        total = total + bill

    print("Total Bill:", total)

    if total >= 5000:
        discount = total * 10 / 100
        total = total - discount
        print("Discount 10%:", discount)
        print("After Discount:", total)

    print("Categories:", categories_set)
    print("Thank you!")
```
